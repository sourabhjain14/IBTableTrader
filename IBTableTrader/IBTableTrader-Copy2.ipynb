```python
import pandas as pd
from ipydatagrid import DataGrid,TextRenderer
from ipywidgets import widgets
```

```python
table_w_ = pd.DataFrame(columns=['Symbol','Bid','Ask','MaxLoss','EntryPrice','StopPrice','Shares',
                                '1stTarget','2ndTarget','StopPct','Value','BUY','SELL','Comment'], index=range(30))
renderers = {
    "MaxLoss": TextRenderer(text_color="black", background_color="red"),
    "EntryPrice": TextRenderer(text_color="black", background_color="green"),
    "StopPrice": TextRenderer(text_color="black", background_color="orange"),
    "BUY": TextRenderer(text_color="black", background_color="green"),
    "SELL": TextRenderer(text_color="black", background_color="red"),
}
dg = DataGrid(table_w_, editable=True,renderers=renderers,layout={"height": "1000px","width":"100%"})
#table_w = dg.data
```

```python
supertabg = widgets.VBox([dg])
supertabg
```
